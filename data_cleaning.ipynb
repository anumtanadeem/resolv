```python
# CLEANINGG

# converting "Date of Purchase" column to datetime format
df["Date of Purchase"] = pd.to_datetime(df["Date of Purchase"]) 

df["Date of Purchase"].dtype    # verifying the conversion
```

```python
# Replacing the product placeholder
df["Ticket Description"] = df.apply(
    lambda row: row["Ticket Description"].replace(
        "{product_purchased}",
        row["Product Purchased"]
    ),
    axis=1
)
df["Ticket Description"].head() # verifying
```

```python
df["Ticket Description"].str.contains(
    r"\{product_purchased\}",
    regex=True,
    na=False
).sum()
```

```python
# convert the response/resolution timestamps to datetime format
df["First Response Time"] = pd.to_datetime(df["First Response Time"])

df["Time to Resolution"] = pd.to_datetime(df["Time to Resolution"])


df[["First Response Time", "Time to Resolution"]].dtypes
```

```python
# Creating an anonymized Customer ID
df["Customer ID"] = (
    df["Customer Email"]
    .astype("category")
    .cat.codes
)  # creates id like 1, 2 ,3

#but we want 0001, 0002, 0003 so
customer_ids = {
    email: f"C{i:04d}"
    for i, email in enumerate(
        df["Customer Email"].unique(),
        start=1
    )
}

df["Customer ID"] = df["Customer Email"].map(customer_ids)

df[["Customer ID", "Customer Email"]].head()
```

```python
# Sorting the DataFrame by Customer ID and Ticket ID
df = df.sort_values(
    ["Customer ID", "Ticket ID"]
).reset_index(drop=True)
```

```python
from sqlalchemy import create_engine

# creates a connection to the MySQL database using SQLAlchemy
engine = create_engine("mysql+pymysql://root:database.777@localhost:3306/customer_support")

df.to_sql(
    "tickets",
    con=engine,
    if_exists="append",   # table already exists with correct schema
    index=False,          # dont save the index as a column in the database
    chunksize=1000        # insert in batches, safer for 8000+ rows
)
```

```python
df.to_csv(
    "data/customer_support_tickets_cleaned.csv",
    index=False,
    quoting=1  # csv.QUOTE_ALL -> wraps every field in quotes, so embedded commas can't break column boundaries
)
print("Cleaned dataset saved successfully.")
print(df.shape)
```

```python
# Sort by email so tickets for the same email are together
check = df.sort_values("Customer Email")

# Keep the columns you want to inspect
check = check[[
    "Customer Email",
    "Customer Name"
]]

# Show every occurrence
print(check.to_string(index=False))
```

```python
# Find emails that occur more than once
duplicate_emails = df[df["Customer Email"].duplicated(keep=False)].copy()

# Sort them together
duplicate_emails = duplicate_emails.sort_values(
    ["Customer Email", "Customer Name"]
)

print(duplicate_emails[["Customer Email", "Customer Name"]].to_string(index=False))
```

```python
name_changes = (
    duplicate_emails
    .groupby("Customer Email")["Customer Name"]
    .agg(lambda x: x.dropna().unique().tolist())
    .reset_index()
)

name_changes["Number of Names"] = name_changes["Customer Name"].apply(len)

# Only emails associated with 2+ different names
name_changes = name_changes[name_changes["Number of Names"] > 1]

print(name_changes.to_string(index=False))
```
